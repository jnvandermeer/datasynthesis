```python
import importlib
import datasynthesis
importlib.reload(datasynthesis)
from datasynthesis import DataSynthesis
from pathlib import Path
import re
import csv # python appaarently has its own csv reader!
```

```python
csv_files = list(Path('../import_these').glob('*.csv'))
```

```python
for i, f in enumerate(csv_files):
    print(f'File {i+1} -> {f}')
```

```python
for i, csv_file in enumerate(csv_files):
    print('\n\n')
    print(f'-----file: {i+1}-------------------')
    print(csv_file)
    print('--------------------------------')

    fields = next(csv.reader(open(csv_file)))
    matches = [field for field in fields if re.match(r'timepoint', field)]
    
    if any(matches):
        print('---> !! this is data with time evolution')
        do_time_evolution = True
    else:
        pass
        do_time_evolution = False
        # print('this is data with no time evolution in it')


    if i+1==9:
        ds = DataSynthesis(max_participants=10000, seed=100)
    else:
        ds = DataSynthesis(max_participants=1000, seed=1)


    my_cols_to_keep=['trial']
    if 'sex' in fields: 
        my_cols_to_keep.append('sex')


    # this is our Melbourne data, we handle this different from anythign else
    if i+1 == 9:
        ds.load(
            csv_file,
            participantid='participant_id',
            data='value',
            type='variable',
            unit='unit',
            cols_to_keep='repeat',
            session='trial',
            time_value=None,
            time_unit=None
        )

    # this is our QUT data; we have timepoint data and normal data, let's do tthat...
    else:
        if do_time_evolution:
            ds.load(
                csv_file,
                participantid='participant_id',
                data='value',
                type='variable',
                unit='unit',
                cols_to_keep=my_cols_to_keep,
                session='period',
                time_value='timepoint_value',
                time_unit='timepoint_unit'
            )
        else:
            ds.load(
                csv_file,
                participantid='participant_id',
                data='value',
                type='variable',
                unit='unit',
                cols_to_keep=my_cols_to_keep,
                session='period',
                time_value=None,
                time_unit=None
            )

    
    

    ds.checkit()

    # if i+1 == 5:
    #     break

    try:
        ds.synthesize()
        ds.export()

        
    except:
        print('DID NOT WORK')
        break


    
    

    
    
```
