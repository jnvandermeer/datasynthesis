```python
# get the information inside:
subject = 'participant_id'
# which columns contain the actual data; the (possibly) variable type (what it is), and the unit (i.e. cm)?
data_to_transform = {'data': 'value', 'type': 'variable', 'unit': 'unit'}
# what to keep as-is:
cols_to_keep = ['sex']
# optional session id:
session_information = 'period'
# optional time id:
time_information = {'values': 'timepoint_value', 'units': 'timepoint_unit'}
```

```python
# we always scramble participant IDs:
old_sub_data =  list(df[subject].unique())
new_sub_data = [f'S{i+1:05d}' for i in range(len(old_sub_data))]
random.shuffle(new_sub_data)
sub_replacement_table = list(zip(old_sub_data, new_sub_data))
```

```python
sub_replacement_table
```

```python
sessions = list(df[session_information].unique())
sessions
```

```python
type_and_unit = df[[data_to_transform['type'], data_to_transform['unit']]].drop_duplicates().values.tolist()
```

```python
for t, u in type_and_unit:
    print(f'{t} {u}')
```

```python
data_to_transform.values()
```

```python
my_conditions = []
```

```python
my_conditions.append((subject, "==", old_sub_data[0]))
my_conditions.append((session_information, "==", sessions[0]))
my_conditions.append((data_to_transform['type'], "==", type_and_unit[0][0]))
my_conditions.append((data_to_transform['unit'], "==", type_and_unit[0][1]))
```

```python
my_conditions
```

```python
cols_to_select = []
cols_to_select.append(subject)
cols_to_select.append(session_information)
cols_to_select.append(data_to_transform['type'])
cols_to_select.append(data_to_transform['unit'])
cols_to_select.append(data_to_transform['data'])
for c in cols_to_keep:
    cols_to_select.append(c)
cols_to_select.append(time_information['values'])
cols_to_select.append(time_information['units'])
```

```python
query = " and ".join(
    f"`{col}` {op} {value!r}"
    for col, op, value in my_conditions
)
```

```python
query
```

```python
df_to_process = df.query(query)[cols_to_select]
df_to_process
```

```python
# so, now we are at the point, where we can extrac the values. Since the participant ID is like actual infromation, and the values are the particpant information.
# so, we can grab the list of the 'value', and do something with that.
# it dpeends on whether it's a single point or a vector at this stage
v = np.array(df_to_process[data_to_transform['data']])
```
